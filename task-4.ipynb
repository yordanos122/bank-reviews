```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud

df = pd.read_csv('/content/data/processed/reviews_with_sentiment.csv')

# Rating distribution per bank
sns.countplot(data=df, x='rating', hue='bank')
plt.savefig('/content/plots/rating_distribution.png')

# Top keywords WordCloud per bank
for bank in df['bank'].unique():
    text = " ".join(df[df['bank']==bank]['review_text'])
    wc = WordCloud(width=800, height=400, background_color='white').generate(text)
    wc.to_file(f'/content/plots/wordcloud_{bank}.png')
```
