```python
# Install required libraries
!pip install pandas google-play-scraper nltk

# Scrape reviews (example for CBE)
from google_play_scraper import reviews
import pandas as pd

cbe_reviews, _ = reviews('com.cbe.mobile', count=400)
df_cbe = pd.DataFrame(cbe_reviews)
df_cbe['bank'] = 'CBE'
df_cbe.to_csv('/content/data/raw/cbe_reviews.csv', index=False)
```

```python
df = pd.read_csv('/content/data/raw/cbe_reviews.csv')
df.drop_duplicates(subset='content', inplace=True)

# Check if 'at' column exists before processing it
if 'at' in df.columns:
    df['review_date'] = pd.to_datetime(df['at']).dt.strftime('%Y-%m-%d')
else:
    print("Warning: 'at' column not found. Skipping review date conversion.")
    df['review_date'] = None # Or handle as appropriate, e.g., create an empty column

df.rename(columns={'content':'review_text','score':'rating'}, inplace=True)
df.to_csv('/content/data/processed/clean_reviews.csv', index=False)
```
